## What is FalconPy?

[FalconPy](https://developer.crowdstrike.com/sdks/python/) is the official CrowdStrike Python SDK.
It provides a Pythonic interface to every CrowdStrike Falcon API service
collection, from Host management to Real Time Response to Vulnerability
Management and beyond.

**Key benefits:**
- Handles OAuth2 token management automatically
- Supports every Falcon API service collection
- Provides both Service Class and Uber Class interfaces
- Fully typed with comprehensive error handling

## Lab Overview

**The case:** A host just threw a privilege-escalation
alert. Notebooks 01-05 walk that incident end to end, live against a
