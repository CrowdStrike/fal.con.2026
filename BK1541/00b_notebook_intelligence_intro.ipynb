# Intro to Notebook Intelligence (AI Assistant)

This lab environment ships with [Notebook Intelligence](https://github.com/notebook-intelligence/notebook-intelligence)
(NBI), a JupyterLab extension that adds an AI chat sidebar and inline code
completion, wired to an internal LLM proxy for this lab.

This feature is optional and instructor-controlled. If you don't see the
icons described below, it simply isn't enabled for this session. Skip
ahead to `01_triage_the_queue.ipynb`.

## Finding it in JupyterLab

- Chat sidebar: look for a chat/assistant icon in the left sidebar.
  Clicking it opens a chat panel alongside your notebook.
- Inline completion: start typing code in a code cell and pause for a
  moment; a greyed-out ("ghost text") suggestion may appear inline.

Both features talk to the same backend model, just through different
interfaces.

## Using the chat panel

The chat panel works like any AI chat interface: type a question, get a
response. A few things worth knowing for this lab specifically:

- You can select code in a cell before asking a question. The assistant
  will typically use that selection as context.
- It has read context on your open notebook, but it cannot act on your
  behalf (more on that below). You stay in control of what actually runs.
- Good first question to try: "What does the `query_devices_by_filter`
  method on FalconPy's `Hosts` class do?"

## Inline code completion

While writing code, NBI can suggest a completion as ghost text right where
your cursor is:

- `Tab`: accept the suggestion
- `Esc`: dismiss it and keep typing your own code

![Tab Completion](data/images/nbi_tab_completion.png)

This works well for boilerplate (imports, loops, `try`/`except` blocks) but
treat it the same way you'd treat autocomplete from any tool: read what it
inserts before you run it.

## Explain, fix, and generate code

While writing code, NBI can explain, fix, and generate code.

- Right click, then choose "Notebook Intelligence" to view the options

![Right Click Menu](data/images/nbi_right_click.png)

This works well for boilerplate (imports, loops, `try`/`except` blocks) but
treat it the same way you'd treat autocomplete from any tool: read what it
inserts before you run it.

## What it can and can't do here

For safety in this shared, multi-tenant lab, this deployment deliberately
disables the assistant's ability to:

- Execute shell commands on your behalf
- Edit files directly without you reviewing and running the change

In other words: it's a chat partner and a completion suggester, not an
autonomous agent. Anything it proposes, you copy into a cell and run
yourself. You'll also only see one model provider available, this lab's
internal proxy, rather than a pick of external providers.

## Try it out

Select the function below (or part of it) and ask the assistant something
like *"What does this function do, step by step?"* or *"How would I test
this?"*

```python
def dedupe_events(events):
    seen = set()
    result = []
    for e in events:
        key = (e.get("device_id"), e.get("timestamp"))
        if key in seen:
            continue
        seen.add(key)
        result.append(e)
    return result
```

## Tips for working with an AI assistant

1. Verify against the docs. The [FalconPy documentation](https://www.falconpy.io/)
   is the source of truth; use the assistant to speed up understanding,
   not replace it.
2. Don't paste secrets into chat. You shouldn't need to; your
   `FALCON_CLIENT_ID`/`FALCON_CLIENT_SECRET` are already in the environment
   for the notebooks to use directly.
3. Small questions get better answers. Ask about one method or one
   error at a time rather than "how do I do the whole lab."
