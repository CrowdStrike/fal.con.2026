## Pattern 1: Explain code you didn't write

Select the function and ask: "What does this do, and what happens if the API call fails?"

```python
import os
from falconpy import Alerts


def get_privilege_escalation_alert_ids(limit=10):
    alerts = Alerts(
        client_id=os.environ.get("FALCON_CLIENT_ID", ""),
        client_secret=os.environ.get("FALCON_CLIENT_SECRET", ""),
    )
    response = alerts.query_alerts_v2(
        limit=limit,
        filter="product:'epp'+tactic:'Privilege Escalation'",
    )
    if response["status_code"] != 200:
        return []
    return response["body"]["resources"]
```
