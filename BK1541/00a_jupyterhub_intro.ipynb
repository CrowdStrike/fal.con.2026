## What's the difference between JupyterHub and JupyterLab?

- **JupyterHub** is the multi-user server behind the scenes. It's what
  authenticated you when you entered your access code, and it's what
  spawned your own private notebook server, completely separate from
  every other attendee's.
- **JupyterLab** is the web-based IDE you're looking at right now: file
  browser, notebook editor, and more, all in one browser tab.

You never interact with JupyterHub directly after logging in. Everything
from here on happens inside JupyterLab.

## A quick tour of the interface

| Area | What it's for |
|---|---|
| File browser (left sidebar, folder icon) | Navigate between notebooks in this lab |
| Table of Contents (left sidebar, list icon) | Jump between headings in the open notebook |
| Main work area | The open notebook(s), each in its own tab |
| Kernel indicator (top-right of a notebook) | Shows whether your Python process is idle, busy, or disconnected |
| Command palette (`Ctrl+Shift+C` / `Cmd+Shift+C`) | Search any JupyterLab action by name |

Cells come in two flavors:
- Markdown cells: text, like this one (double-click to see the raw source)
- Code cells: Python you can run, with output shown directly below

## Running cells

- `Shift + Enter`: run the current cell, move to the next one
- `Ctrl + Enter`: run the current cell, stay put
- `Esc` then `A`/`B`: insert a new cell above / below
- Cells run in the order you run them, not the order they appear on the
  page. If something looks broken, try Run -> Run All Cells from the
  top menu to reset state cleanly.

Try it below.

```python
# Run this cell (Shift+Enter) to see your kernel's working directory and contents
import os

print("Working directory:", os.getcwd())
print()
print("Files here:")
for name in sorted(os.listdir(".")):
    print(f"  {name}")
```

## Kernels & the idle timeout

The kernel is the actual Python process running your code. It's what
holds your variables in memory between cells.

To keep the lab responsive for everyone, idle kernels are automatically
shut down after 15 minutes of inactivity, even if your browser tab is
still open and connected. Kernels that are actively running something are
never culled, only ones sitting idle.

If you come back after a break and your variables are gone (or a cell
errors with a `NameError`), your kernel was recycled. Just use
Run -> Run All Cells to pick back up where you left off.

## Saving your work

- `Ctrl + S` / `Cmd + S` saves the current notebook, same as any editor.
- JupyterLab also autosaves periodically.
- Your notebooks live in your own private workspace and will still be
  there if you get disconnected and log back in during this session.
- That workspace does not survive past the end of the lab. If you want
  to keep anything (code you wrote, notes, output), download it
  (File -> Download) before the session ends.
