```python
# TODO: fill in what you found in each notebook
trigger_aid = None         # the agent_id from notebook 01
trigger_host = None        # the hostname from notebook 01 (WS-DEV-01)
cve_id = None              # the CVE from notebook 02 (the git one)
case_actors = None         # the adversaries from notebook 02 (a list like ["NAME", "NAME"])
exposed_host_count = None  # the number of hosts from notebook 03
fix_action = None          # the action text from notebook 04
```

## Fallback

Run this only if you are missing a value from an earlier notebook.

```python
import json

try:
    with open("data/sample_responses/vulnerabilities_query.json") as handle:
        response = {"status_code": 200, "body": json.load(handle)}
except FileNotFoundError:
    print("sample file not found")
    response = None

if response:
    for vuln in response["body"]["resources"]:
        if "git" in vuln["apps"][0]["product_name_version"].lower():
            cve_id = vuln["cve"]["id"]
            case_actors = vuln["cve"].get("actors") or []
            break

    print("cve:", cve_id)
    print("adversaries:", ", ".join(case_actors))
```

## Write the summary

Run the cell below to see your full incident summary. This is the artifact you ship: it goes
into your ticketing platform (Jira, ServiceNow, etc.) as the triage output, or to leadership
as a one-page brief on what happened, how bad it is, and what needs to happen next.

```python
from rich import print as rprint
from rich.table import Table

table = Table(title="Incident summary")
table.add_column("field", style="cyan")
table.add_column("value")
table.add_row("Host", f"{trigger_host} ({trigger_aid})")
table.add_row("Entry point", "httpd (Apache path traversal)")
table.add_row("Root cause", cve_id)
table.add_row("Attribution", ", ".join(case_actors or []))
table.add_row("Blast radius", f"{exposed_host_count} hosts")
table.add_row("Fix", fix_action)
rprint(table)
```
